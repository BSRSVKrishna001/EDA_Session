## Head Map

- Any Matrix we can visualize in Heat Map

- Heat Map will provide colors for different values

- Heat Map also provide color bar which indicates, the color and its value

- For example the value range from 0.8 to 1 display as **Blue Color**

- So we no need to check values, we can directly see the Blue color

- So immediately we can sence Blue color means highest value which is 0.8 to 1

- Heat Map is available in **seaborn**
