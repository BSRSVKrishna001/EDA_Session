```python
## Step-1: Take the prevailing_wage data
## Step2: cal mean of the data
## Step3: cal std of the data
## Step4: Nr=data=mean (step1- step2)
## Step5: Z_Wage_Data=Nr/std= Step4/Step3
## Step6: Create a new column visa_df['Z_Wage_Data']=Z_Wage
```

```python
df=visa_df['prevailing_wage']
Mean=df.mean()
Std=df.std()
Nr=df-Mean
Z_Wage_Data=Nr/Std
visa_df['Z_Wage_Data']=Z_Wage_Data
```

```python
visa_df[['prevailing_wage','Z_Wage_Data']]
```

```python
plt.Figure(figsize=(20,6))
plt.suptitle('Raw-Data Vs Z-Data')
# plt.Figure(figsize=(10,6))
plt.subplot(2,2,1).hist(visa_df['prevailing_wage'])
plt.title('prevailing_wage')
plt.subplot(2,2,2).hist(visa_df['Z_Wage_Data'])
plt.title('Z_Wage_Data')
plt.subplot(2,2,3).boxplot(visa_df['prevailing_wage'],vert=False)
plt.subplot(2,2,4).boxplot(visa_df['Z_Wage_Data'],vert=False)
plt.show()
```

# **Min Max Scaler**

- X scaler=X-Xmin/Xmax-Xmin

formula Image

```python
## Step1: Read the Data
## Step2: cal min
## Step3: cal max
## Step4: nr=step1-step2
## Step5: Dr=step3-step2
##Step6: nr/Dr
## Step7: visa_df['wade_min_max']=nr/dr
```

```python
df=visa_df['prevailing_wage']
Min=df.min()
Max=df.max()
nr=df-Min
dr=Max-Min
col=nr/dr
visa_df['Wage_Min_Max']=col
```
